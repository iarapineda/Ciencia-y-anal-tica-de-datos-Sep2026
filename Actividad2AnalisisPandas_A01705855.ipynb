8. Utiliza el dataframe de indicadores diarios para crear un nuevo dataframe `indicadores_mensuales`. Para ello:
- Crea una nueva columna `mes` a partir de la fecha que se encuentra en el índice.
- Haciendo uso de `groupby()`, calcula los promedios mensuales y conserva únicamente las columnas `humedad_relativa`, `temperatura_celsius` y `velocidad_viento` en el nuevo dataframe.

```python
indicadores_diarios["mes"] = pd.to_datetime(
    indicadores_diarios.index
).month

#promedios
indicadores_mensuales = indicadores_diarios.groupby("mes")[
    ["humedad_relativa", "temperatura_celsius", "velocidad_viento"]
].mean()

display(indicadores_mensuales)
display(indicadores_mensuales.shape)
```

**Respuesta a Pregunta 8**

Creé la columna mes a partir de las fechas del índice de indicadores_diarios.Usé groupby() para calcular los promedios mensuales de los máximos diarios de humedad relativa, temperatura y velocidad del viento. El df indicadores_mensuales contiene doce registros, uno por mes, y únicamente los tres indicadores solicitados.

Agosto = mayor promedio de temperaturas máximas diarias

Septiembre = mayor promedio de humedad relativa máxima diaria

Febrero = velocidad máxima diaria del viento

9. Construye el mismo dataframe mensual, pero esta vez utilizando la función `pivot_table()`.
- Añade la columna `indice_calor` usando la fórmula de Thom (*Temperature - Humidity Index*, TDI):
> `temperatura_celsius + 0.33 x humedad_relativa - 0.7 x velocidad_viento - 4`

```python
indicadores_mensuales = indicadores_diarios.pivot_table(
    index="mes",
    values=[
        "humedad_relativa",
        "temperatura_celsius",
        "velocidad_viento"
    ],
    aggfunc="mean"
)

indicadores_mensuales = indicadores_mensuales[
    ["humedad_relativa", "temperatura_celsius", "velocidad_viento"]
]

display(indicadores_mensuales)

#indice de calor IC=T+0.33H−0.7V−4
#T: promedio mensual de las temperaturas máximas diarias.
#H: promedio mensual de las humedades relativas máximas diarias.
#V: promedio mensual de las velocidades máximas diarias del viento.

indicadores_mensuales["indice_calor"] = (
    indicadores_mensuales["temperatura_celsius"]
    + 0.33 * indicadores_mensuales["humedad_relativa"]
    - 0.7 * indicadores_mensuales["velocidad_viento"]
    - 4
)

#df final
display(indicadores_mensuales)
display(indicadores_mensuales.shape)
```

**Respuesta a Pregunta 9.**

El df final tiene 12 registros y 4 columnas. pivot_table() se usa para calcular los promedios mensuales de humedad relativa, temperatura y velocidad del viento a partir de los máximos diarios. Se obtuvo la misma estructura y los mismos resultados que mediante groupby(), y después se incorporó la columna indice_calor, aplicando la fórmula proporcionada en la actividad a los tres indicadores mensuales.

10. Recuerda que de las cuatro funciones estudiadas para manipular la estructura del dataframe:
- `melt/pivot` hacen las transformaciones de manera controlada, es decir puedes
definir por medio de sus parámetros qué variables quedarán como índices, columnas y valores.
- `stack/unstack` la conversión se aplica siempre sobre los niveles inferiores de index/columns.
- Convierte el dataframe del ejercicio anterior a formato largo usando: `melt()` y `stack()` y comenta las diferencias.

```python
#formato largo melt
mensuales_melt = indicadores_mensuales.reset_index().melt(
    id_vars="mes",
    var_name="indicador",
    value_name="valor"
)

display(mensuales_melt)
```

```python
#formato largo stack
mensuales_stack = indicadores_mensuales.stack()

display(mensuales_stack)
```

**Respuesta a Pregunta 10.**

Ambas transformaciones conservan la información original, pero melt() permite especificar las columnas identificadoras y los nombres de las variables resultantes, mientras que stack() traslada los niveles de columnas al índice.

Con melt(), se obtiene un DF de 48 registros y tres columnas: mes, indicador y valor. Por su parte, stack() genera una Series con 48 valores organizados mediante un índice multinivel.
